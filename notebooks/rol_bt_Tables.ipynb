# Role table and Betweenness table

```python
import pickle
import pandas as pd
import networkx as nx
```

```python
l_g_file=['carnival_tourist','no_carnival_tourist','carnival_residents','no_carnival_residents']
```

## Role table

```python
list_rol=list()
for i in l_g_file:
    with open('../data/'+i+'_rol.pickle', 'rb') as handle:
        list_rol.append( pickle.load(handle))
        
df = pd.DataFrame({
    l_g_file[0]: pd.Series(list_rol[0]),
    l_g_file[1]: pd.Series(list_rol[1]),
    l_g_file[2]: pd.Series(list_rol[2]),
    l_g_file[3]: pd.Series(list_rol[3])
})
df       
```

## Betweenness table

```python
list_btw=list()
for i in l_g_file:
    G=nx.read_graphml('../data/'+i+'.graphml')
    
for i in G.nodes():
    btw=nx.betweenness_centrality(G, k=None, normalized=True, weight=None, endpoints=False, seed=None)
    list_btw.append(btw)
df = pd.DataFrame({
    l_g_file[0]: pd.Series(list_btw[0]),
    l_g_file[1]: pd.Series(list_btw[1]),
    l_g_file[2]: pd.Series(list_btw[2]),
    l_g_file[3]: pd.Series(list_btw[3])
})
df
```

# Select the configuration by minimising the Haming distance.

```python
import sys
sys.path.append('../bin')

import networkx as nx

import pandas as pd
import numpy as np
from scipy.spatial.distance import hamming
import random
import guimera as gm

g_files = ["no_carnival_tourist","no_carnival_residents","carnival_residents","carnival_tourist"]
```

## Calculate Haming distance between columns

We calculate the Haming distance between the columns of the dataframe, each of which represents a guimera execution.
if there's a tie, a random column is selected


```python
for n_file in g_files:
    
    df_community = pd.read_csv(f"../models/{n_file}_guimera_repeated_cm.csv")

    df_role = pd.read_csv(f"../models/{n_file}_guimera_repeated_role.csv")
    df_reference_role = pd.read_csv(f"../models/{n_file}_guimera_repeated_role_mean.csv")
    
    # Skip the first column (identifiers)
    data = df_role.iloc[:, 1:]
    reference = df_reference_role.iloc[:, 1].values  # Assuming the reference roles are in the second column

    cols = data.columns
    n = len(cols)
    dist_list = []

    for i in range(n):
        dist = hamming(data.iloc[:, i], reference)
        dist_list.append(dist)

    # Create a DataFrame of results
    results = pd.DataFrame({
        'Column': cols,
        'distance_to_reference': dist_list
    }).sort_values('distance_to_reference')

    # Select the column most similar to the reference
    min_distance = results['distance_to_reference'].min()
    min_cols = results[results['distance_to_reference'] == min_distance]['Column'].tolist()
    selected_column = random.choice(min_cols)

    print("Hamming distances to reference:")
    print(results)
    print("\nColumn most similar to the reference selected:")
    print(selected_column)
    # We save the settings closest to the average for later representation. 
    df_z=pd.read_csv(f"../models/{n_file}_guimera_repeated_z.csv")
    df_p=pd.read_csv(f"../models/{n_file}_guimera_repeated_p.csv")
    
 
    dc_community=df_community.set_index('Unnamed: 0')[selected_column].to_dict()
    cm=list()
    for i in range(0,max(dc_community.values())+1):
        cm.append(list())
        for j in  dc_community:
           if dc_community[j]==i:
               cm[-1].append(j)
        
    gm.to_pickle(df_z.set_index('Unnamed: 0')[selected_column].to_dict(),df_p.set_index('Unnamed: 0')[selected_column].to_dict(),df_role.set_index('Unnamed: 0')[selected_column].to_dict(),cm,n_file)

    
    
```
